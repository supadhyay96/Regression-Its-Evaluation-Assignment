    **Q1. What is Simple Linear Regression?**

        -> ### **Simple Linear Regression**

    **Simple Linear Regression** is a supervised machine learning algorithm used to find the linear relationship between two variables: an independent variable (X) and a dependent variable (Y). It predicts the value of the dependent variable based on the value of the independent variable by fitting a straight line to the data.

    **Mathematical equation:**

    \\[ Y = b_0 + b_1X \\]

    Where:

    - **Y** – Predicted dependent variable
    - **X** – Independent variable (input)
    - **b₀** – Intercept (value of Y when X = 0)
    - **b₁** – Slope (change in Y for a one-unit increase in X)

    **Example:** If we want to predict a student's marks based on the number of hours studied, Simple Linear Regression can help estimate the marks. Generally, as study hours increase, marks may also increase.

    \\(\hat{y} = b_0 + b_1x\\)

    \\(\hat{y} = \text{8.49} - \text{0.54}x\\)

    Estimated intercept · R-squared = 0.72

    **Applications:**

    - Predicting sales based on advertising expenditure.
    - Estimating salary based on years of experience.
    - Predicting house prices based on area.
    - Analyzing the relationship between two variables.

    **In short:** Simple Linear Regression helps understand the relationship between two variables and predict numerical outcomes using a straight-line equation.

    **Q2.  What are the key assumptions of Simple Linear Regression?**

        -> ### **Key Assumptions of Simple Linear Regression**

    Simple Linear Regression relies on the following assumptions to produce reliable predictions:

    1\. **Linearity:** There should be a linear relationship between the independent variable (X) and the dependent variable (Y).

    2\. **Independence:** The observations should be independent of each other; one observation should not influence another.

    3\. **Homoscedasticity (Constant Variance):** The residuals (errors) should have a constant variance across all values of X.

    4\. **Normality of Residuals:** The residuals should be approximately normally distributed, especially for hypothesis testing and confidence intervals.

    5\. **No Significant Outliers:** Extreme outliers should not excessively influence the regression line or distort the relationship between variables.

    6\. **Correct Model Specification:** The model should include the appropriate variables and represent the relationship correctly.

    **In short:** These assumptions ensure that the regression model provides accurate, unbiased, and reliable results.

    **Q3.  What is heteroscedasticity, and why is it important to address in regression models?**

        ->

    **Heteroscedasticity** occurs in a regression model when the variance of the residuals (errors) is not constant across different values of the independent variable. This means the spread of errors changes as the predicted values increase or decrease.

    Ideally, regression residuals should have constant variance, which is known as **homoscedasticity**.

    ### **Why is it important to address?**

    - **Unreliable standard errors:** It can make standard errors inaccurate.
    - **Incorrect hypothesis testing:** It may lead to incorrect p-values and conclusions about the significance of variables.
    - **Reduced confidence in results:** Confidence intervals may become unreliable.
    - **Poor model interpretation:** It can make it difficult to draw accurate conclusions from the regression model.

    ### **Example**

    Suppose we predict house prices based on house size. For smaller houses, prediction errors may be small, but for larger houses, the errors may be much greater. This changing spread of errors indicates heteroscedasticity.

    \\(\hat{y} = b_0 + b_1x\\)

    \\(\hat{y} = \text{8.49} - \text{0.54}x\\)

    Estimated intercept · R-squared = 0.72


    **In short:** Heteroscedasticity means the error variance is not constant. Addressing it helps ensure reliable statistical inference and better interpretation of regression results.


    **Q4. What is Multiple Linear Regression?**

        -> ### **Multiple Linear Regression**

    **Multiple Linear Regression** is a supervised machine learning algorithm used to predict the value of a dependent variable based on two or more independent variables. It establishes a linear relationship between the target variable and multiple input variables.

    **Mathematical Equation:**

    \\[ Y=b_0+b_1X_1+b_2X_2+\cdots+b_nX_n \\]

    Where:

    - **Y** – Dependent (target) variable
    - **X₁, X₂, ..., Xₙ** – Independent (input) variables
    - **b₀** – Intercept
    - **b₁, b₂, ..., bₙ** – Coefficients representing the effect of each independent variable

    **Example:** Predicting house prices based on area, number of bedrooms, location, and age of the house.

    **Applications:**

    - Predicting house prices using multiple features.
    - Forecasting sales using price, advertising, and demand.
    - Predicting salary based on experience, education, and skills.
    - Estimating business revenue using multiple factors.

    **In short:** Multiple Linear Regression predicts a numerical outcome using multiple independent variables by fitting a linear equation to the data.

    Q5. What is polynomial regression, and how does it differ from linear
    regression?

        -> ### **Polynomial Regression**

    **Polynomial Regression** is a supervised machine learning algorithm used to model a non-linear relationship between independent and dependent variables. It uses polynomial equations (powers of X) to fit a curved line to the data.

    **Mathematical Equation:**

    \\[ Y=b_0+b_1X+b_2X^2+\cdots+b_nX^n \\]

    Where:

    - **Y** – Dependent variable.
    - **X** – Independent variable.
    - **b₀** – Intercept.
    - **b₁, b₂, ...** – Coefficients.
    - **n** – Degree of the polynomial.

    ### **Difference Between Linear and Polynomial Regression**

    | Linear Regression                    | Polynomial Regression                       |
    | ------------------------------------ | ------------------------------------------- |
    | Models a straight-line relationship. | Models a curved relationship.               |
    | Uses the first-degree equation.      | Uses higher-degree terms such as X² and X³. |
    | Suitable for linear relationships.   | Suitable for non-linear relationships.      |
    | Simple to understand and interpret.  | More complex to interpret.                  |

    **Example:** Predicting salary based on experience when salary increases at a changing rate rather than a constant rate.

    **Applications:**

    - Predicting population growth.
    - Forecasting sales trends.
    - Analyzing temperature changes.
    - Predicting business growth.

    **In short:** Linear Regression fits a straight line, whereas Polynomial Regression fits a curve to capture more complex relationships between variables.

```python
#Question 6: Implement a Python program to fit a Simple Linear Regression model to
#the following sample data:
#● X = [1, 2, 3, 4, 5]
#● Y = [2.1, 4.3, 6.1, 7.9, 10.2]
#Plot the regression line over the data points.


import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression


X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
Y = np.array([2.1, 4.3, 6.1, 7.9, 10.2])


model = LinearRegression()
model.fit(X, Y)

Y_pred = model.predict(X)


print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)

plt.scatter(X, Y, color='blue', label='Actual Data')
plt.plot(X, Y_pred, color='red', label='Regression Line')

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Simple Linear Regression")
plt.legend()
plt.show()
```

```python
# Q7. Fit a Multiple Linear Regression model on this sample data:
#● Area = [1200, 1500, 1800, 2000]
#● Rooms = [2, 3, 3, 4]
#● Price = [250000, 300000, 320000, 370000]
#Check for multicollinearity using VIF and report the results.


import pandas as pd
from sklearn.linear_model import LinearRegression
from statsmodels.stats.outliers_influence import variance_inflation_factor

data = {
    'Area': [1200, 1500, 1800, 2000],
    'Rooms': [2, 3, 3, 4],
    'Price': [250000, 300000, 320000, 370000]
}

df = pd.DataFrame(data)


X = df[['Area', 'Rooms']]
Y = df['Price']


model = LinearRegression()
model.fit(X, Y)

Y_pred = model.predict(X)

print("Coefficients:", model.coef_)
print("Intercept:", model.intercept_)
print("Predicted Prices:", Y_pred)


vif_data = pd.DataFrame()
vif_data['Feature'] = X.columns
vif_data['VIF'] = [
    variance_inflation_factor(X.values, i)
    for i in range(X.shape[1])
]

print("\nVIF Results:")
print(vif_data)
```

```python
# Q8  Implement polynomial regression on the following data:
#● X = [1, 2, 3, 4, 5]
#● Y = [2.2, 4.8, 7.5, 11.2, 14.7]
#Fit a 2nd-degree polynomial and plot the resulting curve.


import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression


X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
Y = np.array([2.2, 4.8, 7.5, 11.2, 14.7])


poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)


model = LinearRegression()
model.fit(X_poly, Y)


X_curve = np.linspace(1, 5, 100).reshape(-1, 1)
X_curve_poly = poly.transform(X_curve)
Y_curve = model.predict(X_curve_poly)


print("Coefficients:", model.coef_)
print("Intercept:", model.intercept_)

plt.scatter(X, Y, color='blue', label='Actual Data')
plt.plot(X_curve, Y_curve, color='red', label='Polynomial Curve')

plt.xlabel("X")
plt.ylabel("Y")
plt.title("2nd Degree Polynomial Regression")
plt.legend()
plt.show()
```

```python
# Create a residuals plot for a regression model trained on this data:
#● X = [10, 20, 30, 40, 50]
#● Y = [15, 35, 40, 50, 65]
#Assess heteroscedasticity by examining the spread of residuals.


import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression


X = np.array([10, 20, 30, 40, 50]).reshape(-1, 1)
Y = np.array([15, 35, 40, 50, 65])


model = LinearRegression()
model.fit(X, Y)


Y_pred = model.predict(X)


residuals = Y - Y_pred

print("Predicted Values:", Y_pred)
print("Residuals:", residuals)


plt.scatter(Y_pred, residuals, color='blue')
plt.axhline(y=0, color='red', linestyle='--')

plt.xlabel("Predicted Values")
plt.ylabel("Residuals")
plt.title("Residual Plot")
plt.show()
```

    Q10. Imagine you are a data scientist working for a real estate company. You
    need to predict house prices using features like area, number of rooms, and location.
    However, you detect heteroscedasticity and multicollinearity in your regression
    model. Explain the steps you would take to address these issues and ensure a robust
    model.

        -> ### **Steps to Address Heteroscedasticity and Multicollinearity in House Price Prediction**

    As a data scientist working for a real estate company, I would take the following steps to improve the accuracy and reliability of the house price prediction model.

    **1. Identify Heteroscedasticity**

    - Create a residual plot to check whether the variance of errors is constant.
    - If residuals show an increasing or decreasing spread, heteroscedasticity may be present.

    **2. Handle Heteroscedasticity**

    - Apply a logarithmic transformation to house prices to reduce unequal variance.
    - Use Weighted Least Squares (WLS) regression to give different weights to observations.
    - Use robust standard errors to make statistical results more reliable.

    **3. Identify Multicollinearity**

    - Calculate Variance Inflation Factor (VIF) for independent variables.
    - Check the correlation between features such as area and number of rooms.
    - A VIF value above 5 may indicate high multicollinearity.

    **4. Handle Multicollinearity**

    - Remove highly correlated or unnecessary features.
    - Combine related features if appropriate.
    - Use Ridge or Lasso Regression to reduce the impact of multicollinearity.
    - Apply feature selection to identify the most important variables.

    **5. Improve the Model**

    - Encode categorical features such as location using One-Hot Encoding.
    - Split the dataset into training and testing sets.
    - Apply feature scaling where required.
    - Train and compare different regression models.

    **6. Evaluate Model Performance**

    Use the following evaluation metrics:

    - **R² Score:** Measures how well the model explains price variations.
    - **MAE:** Measures average absolute prediction errors.
    - **MSE:** Measures average squared prediction errors.
    - **RMSE:** Measures the typical prediction error in price units.

    **7. Validate the Model**

    - Use cross-validation to check model performance.
    - Recheck residual plots and VIF values.
    - Test the model on unseen house price data.

    ### **Conclusion**

    By identifying and handling heteroscedasticity and multicollinearity, applying suitable transformations and regularization techniques, and evaluating the model properly, I can build a more reliable and robust house price prediction model.
